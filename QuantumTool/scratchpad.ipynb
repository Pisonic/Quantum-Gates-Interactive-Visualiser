## Single Qubit 

```python
import qiskit
from qiskit.quantum_info import Statevector
from qiskit import QuantumCircuit
from qiskit.visualization.state_visualization import state_drawer
import sympy
```

```python
q = QuantumCircuit(1)

q.x(0)

state = Statevector.from_instruction(q)
display(state.draw("latex"))
```

## Double Qubit

```python
q = QuantumCircuit(2)

q.h(0)
q.cx(0, 1)
q.z(0)
q.x(1)
q.h(0)
state = Statevector.from_instruction(q)
display(state.draw('latex'))
```

```python
q = QuantumCircuit(2)

q.h(0)
q.cx(0, 1)
q.z(0)
q.x(1)
q.h(0)
q.measure_all()
q.draw('text')
```

```python
q = QuantumCircuit(2)

q.h(0)
q.cx(0, 1)
q.z(0)
q.x(1)
q.h(0)

state = Statevector.from_instruction(q)
counts = state.sample_counts(shots = 1000)
dict(counts)
```
